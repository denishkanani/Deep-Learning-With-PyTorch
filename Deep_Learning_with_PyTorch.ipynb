## Simple Neural Network

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

# Create a Model Class that inherits nn.module
class Model(nn.Module):
  def __init__(self,in_features=4,h1=8,h2=9,out_features=3):
    super().__init__()
    self.fc1 = nn.Linear(in_features,h1)
    self.fc2 = nn.Linear(h1,h2)
    self.out = nn.Linear(h2,out_features)
  def forward(self, X):
    x = F.relu(self.fc1(X)) # Changed x to X here
    x = F.relu(self.fc2(x))
    x = self.out(x)
    return x
```

```python
# Pick a manual Seed for randomization
torch.manual_seed(13)
# Create an instance of the model
model = Model()
```

  ## Real Data to Work

```python
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
url = 'https://gist.githubusercontent.com/curran/a08a1080b88344b0c8a7/raw/0e7a9b0a5d22642a06d3d5b9bcbad9890c8ee534/iris.csv'
my_df = pd.read_csv(url)
```

```python
my_df.head()
```

```python
my_df.tail()
```

```python
# Change last column From strings to integers
my_df['species'] = my_df['species'].replace('setosa', 0.0)
my_df['species'] = my_df['species'].replace('versicolor', 0.0)
my_df['species'] = my_df['species'].replace('virginica', 0.0)
my_df
```

```python
# Train Test Split! Set X,y
X = my_df.drop('species',axis=1).values
y = my_df['species'].values
```

```python
# Convert these to numpy arrays
```

```python
from sklearn.model_selection import train_test_split

# Example demonstrating model.parameters()
# This returns an iterator over the learnable parameters of the model.
# Each parameter is a tensor.
# You typically call model.parameters() when passing them to an optimizer.
# For example, in cell dCALJyjOK60r you have: optimizer = torch.optim.Adam(model.parameters(),lr=0.01)

print("Model Parameters:")
for name, param in model.named_parameters():
    if param.requires_grad:
        print(f"Parameter name: {name}, Shape: {param.shape}, Requires Grad: {param.requires_grad}")
```

```python
# Train Test Split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=40)
```

```python
# Convert X features to float tensors
X_train = torch.FloatTensor(X_train)
X_test = torch.FloatTensor(X_test)
```

```python
# Convert y Labels to tensors long
y_train = torch.LongTensor(y_train)
y_test = torch.LongTensor(y_test)
```

```python
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(),lr=0.01)
```

```python
# Train Model
epochs = 100
losses = []
for i in range(epochs):
  y_pred = model.forward(X_train)
  loss = criterion(y_pred,y_train)
  losses.append(loss)
  if i%10 == 0:
    print(f'Epoch: {i} Loss: {loss}')
  optimizer.zero_grad()
  loss.backward()
  optimizer.step()
```

```python
# Graph it out!
plt.plot(range(epochs), [loss.item() for loss in losses])
plt.ylabel('Loss')
plt.xlabel('Epoch')
```

```python
# Evaluate Model
with torch.no_grad():
  y_eval = model.forward(X_test)
  loss = criterion(y_eval,y_test)
```

```python
loss
```

```python
correct = 0
with torch.no_grad():
  for i, data in enumerate(X_test):
    y_val = model.forward(data)

    if y_test[i] == 0:
      x = "Sentosa"
    elif y_test[i] == 1:
      x = "Versicolor"
    else:
      x = "Virginica"

    print(f'{i+1}.) {str(y_val)} \t {y_test[i]} \t {y_val.argmax().item()}')

    #correct or not
    if y_val.argmax().item() == y_test[i]:
      correct +=1
print(f'We got {correct} Correct!')
```
